## Scenario 01: University Course Registration System

A university wants to manage course registrations for students. Each student can register for multiple courses, but duplicate registrations must not be allowed.

**Your task:**
- Ask for the student's name and the courses they want to register for.
- Store the registered courses in a **list**.
- Use a **set** to identify and prevent duplicate course registrations.
- Store each course's code and credit hours in a **tuple**.
- Use a **dictionary** to maintain course details, including course codes and credit hours.
- Display the final registered courses, total credit hours, and any duplicate entries rejected.

```python
# ─────────────────────────────────────────────────────────────────────────────
# Scenario 01 – University Course Registration System
# Data structures used:
#   • list        → ordered list of successfully registered courses
#   • set         → tracks already-registered course codes (prevents duplicates)
#   • tuple       → immutable (course_code, credit_hours) pair per course
#   • dictionary  → maps course_code → (course_name, credit_hours) as full info
# ─────────────────────────────────────────────────────────────────────────────

# ── Course catalogue: all available courses at the university ──────────────
# key  : course code (string)
# value: tuple → (course_name, credit_hours)
course_catalogue = {
    "CS101": ("Introduction to Programming", 3),
    "MA201": ("Calculus II", 4),
    "PH101": ("Physics Fundamentals", 3),
    "CS202": ("Data Structures", 3),
    "EN101": ("English Communication", 2),
    "DA301": ("Data Analytics", 3),
}

def register_courses():
    """Interactive course registration for one student."""

    student_name = input("Enter student name: ").strip()
    if not student_name:
        print("❌ Student name cannot be empty.")
        return

    registered_courses = []   # LIST  – ordered successful registrations
    registered_codes   = set() # SET   – fast duplicate-check store
    rejected_duplicates = []   # LIST  – duplicate attempts recorded

    print("\nAvailable courses:")
    for code, (name, credits) in course_catalogue.items():
        print(f"  {code} – {name} ({credits} credits)")

    print("\nEnter course codes one by one. Type 'done' or press Escape when finished.")

    while True:
        user_input = input("Course code: ").strip().upper()

        if user_input == "DONE" or user_input == "":  # empty string = Escape key
            break

        if user_input not in course_catalogue:
            print(f"  ⚠ '{user_input}' is not a valid course code. Try again.")
            continue

        if user_input in registered_codes:          # duplicate check via SET
            print(f"  ⚠ '{user_input}' already registered – duplicate rejected.")
            rejected_duplicates.append(user_input)
            continue

        # Valid, non-duplicate course → register it
        registered_codes.add(user_input)           # update SET
        course_name, credit_hours = course_catalogue[user_input]
        course_record = (user_input, credit_hours)  # TUPLE (code, credits)
        registered_courses.append(course_record)    # append to LIST
        print(f"  ✔ Registered: {course_name} ({credit_hours} credits)")

    # ── Summary ──────────────────────────────────────────────────────────────
    total_credits = sum(credits for _, credits in registered_courses)

    print("\n" + "="*55)
    print(f" Registration Summary for: {student_name}")
    print("="*55)

    if registered_courses:
        print(f"{'Course Code':<12} {'Course Name':<30} {'Credits'}")
        print("-"*55)
        for code, credits in registered_courses:
            name = course_catalogue[code][0]
            print(f"{code:<12} {name:<30} {credits}")
        print("-"*55)
        print(f"{'Total Credit Hours:':<42} {total_credits}")
    else:
        print("  No courses registered.")

    if rejected_duplicates:
        print(f"\nDuplicate registrations rejected: {rejected_duplicates}")
    else:
        print("\nNo duplicate registrations attempted.")

# ── Run ────────────────────────────────────────────────────────────────────
register_courses()


```

## Scenario 02: Inventory and Stock Management System

A small shop needs a program to monitor its products, prices, and available stock.

**Your task:**
- Allow the shopkeeper to enter product names, prices, and quantities.
- Use a **dictionary** to store product information, with product names as keys.
- Store each product's price and quantity together in a **tuple**.
- Maintain a **set** of unique product categories.
- Use a **list** to record products that need restocking when their quantity falls below five.
- Display the complete inventory, total inventory value, unique categories, and restocking list.
- Allow the shopkeeper to update stock quantities and prices.

```python
# ─────────────────────────────────────────────────────────────────────────────
# Scenario 02 – Inventory and Stock Management System
# Data structures used:
#   • dictionary  → product_name → {"price_qty": tuple, "category": str}
#   • tuple       → (price, quantity) stored as the value per product
#   • set         → unique categories across all products
#   • list        → products needing restock (qty < 5)
# ─────────────────────────────────────────────────────────────────────────────

RESTOCK_THRESHOLD = 5

# Pre-defined list of available categories to choose from
AVAILABLE_CATEGORIES = ["Fruits", "Dairy", "Bakery", "Grains", "Stationery", "Beverages", "Snacks"]

# Initial inventory dictionary (normalized title case)
inventory = {
    "Apple":    {"price_qty": (30.0,  50), "category": "Fruits"},
    "Milk":     {"price_qty": (55.0,   3), "category": "Dairy"},
    "Bread":    {"price_qty": (45.0,   2), "category": "Bakery"},
    "Cheese":   {"price_qty": (200.0,  4), "category": "Dairy"},
    "Rice":     {"price_qty": (60.0,  25), "category": "Grains"},
}

def get_categories(inv):
    """Return a SET of all unique categories in current inventory."""
    return {details["category"] for details in inv.values()}

def get_restock_list(inv):
    """Return a LIST of product names where quantity falls below RESTOCK_THRESHOLD."""
    return [name for name, details in inv.items()
            if details["price_qty"][1] < RESTOCK_THRESHOLD]

def get_total_value(inv):
    """Calculate total inventory value = sum(price * quantity)."""
    return sum(details["price_qty"][0] * details["price_qty"][1]
               for details in inv.values())

def display_inventory(inv, shopkeeper_name=""):
    """Pretty-print the full inventory."""
    categories   = get_categories(inv)      # SET
    restock_list = get_restock_list(inv)    # LIST
    total_value  = get_total_value(inv)     # float

    header = f" Inventory of {shopkeeper_name}'s Shop" if shopkeeper_name else " Current Inventory"
    print("\n" + "="*68)
    print(header)
    print("="*68)
    print(f" {'Product':<15} {'Category':<14} {'Price (₹)':>10} {'Qty':>6} {'Value (₹)':>12}")
    print("-" * 68)
    for name, details in inv.items():
        price, qty = details["price_qty"]   # unpack TUPLE
        cat        = details["category"]
        value      = price * qty
        flag       = " ⚠ RESTOCK" if qty < RESTOCK_THRESHOLD else ""
        print(f" {name:<15} {cat:<14} {price:>10.2f} {qty:>6} {value:>12.2f}{flag}")
    print("="*68)
    print(f" Total Inventory Value: ₹{total_value:,.2f}")
    print(f" Unique Categories    : {categories}")
    print(f" Restock Needed       : {restock_list if restock_list else 'None'}")

def add_product(inv, name, price, qty, category):
    """
    Add or update a product in inventory.
    If the product already exists:
      - Adds the incoming qty to existing stock
      - Updates the price to the latest price
    """
    if price < 0 or qty < 0:
        print("❌ Price and quantity cannot be negative.")
        return False
    
    # Normalize product name so 'apple' and 'Apple' refer to the same item
    clean_name = name.strip().title()
    
    if clean_name in inv:
        old_price, old_qty = inv[clean_name]["price_qty"]
        new_qty = old_qty + qty
        # Store updated price and combined quantity in a new TUPLE
        inv[clean_name] = {"price_qty": (price, new_qty), "category": category}
        print(f"✔ Existing product '{clean_name}' updated! Added {qty} units (Total: {new_qty}), Price: ₹{price:.2f}.")
    else:
        inv[clean_name] = {"price_qty": (price, qty), "category": category}
        print(f"✔ New product '{clean_name}' successfully added (Qty: {qty}, Price: ₹{price:.2f}).")
    return True

def update_stock(inv, name, new_qty):
    """Update stock quantity of an existing product."""
    clean_name = name.strip().title()
    if clean_name not in inv:
        print(f"❌ Product '{clean_name}' not found.")
        return
    if new_qty < 0:
        print("❌ Quantity cannot be negative.")
        return
    old_price, _ = inv[clean_name]["price_qty"]
    inv[clean_name]["price_qty"] = (old_price, new_qty)  # new TUPLE
    print(f"✔ '{clean_name}' stock updated to {new_qty}.")

def update_price(inv, name, new_price):
    """Update price of an existing product."""
    clean_name = name.strip().title()
    if clean_name not in inv:
        print(f"❌ Product '{clean_name}' not found.")
        return
    if new_price < 0:
        print("❌ Price cannot be negative.")
        return
    _, old_qty = inv[clean_name]["price_qty"]
    inv[clean_name]["price_qty"] = (new_price, old_qty)  # new TUPLE
    print(f"✔ '{clean_name}' price updated to ₹{new_price:.2f}.")

def shopkeeper_portal():
    """Interactive portal for the shopkeeper."""
    shopkeeper = input("Enter Shopkeeper Name (or press Enter/Escape to skip): ").strip()
    if not shopkeeper:
        shopkeeper = "Shopkeeper"

    print(f"\nWelcome, {shopkeeper}!")
    display_inventory(inventory, shopkeeper)

    print("\n--- Add / Restock Products in Inventory ---")
    print("(Type 'done' or press Enter/Escape on product name when finished)")

    while True:
        prod_name = input("\nEnter Product Name: ").strip()
        if not prod_name or prod_name.upper() == "DONE":
            break

        # Input and validate price
        try:
            price_input = input(f"Enter price for '{prod_name}': ").strip()
            if not price_input:
                break
            price = float(price_input)
            if price < 0:
                print("❌ Price cannot be negative.")
                continue
        except ValueError:
            print("❌ Invalid price. Must be a number.")
            continue

        # Input and validate quantity
        try:
            qty_input = input(f"Enter quantity to add for '{prod_name}': ").strip()
            if not qty_input:
                break
            qty = int(qty_input)
            if qty < 0:
                print("❌ Quantity cannot be negative.")
                continue
        except ValueError:
            print("❌ Invalid quantity. Must be an integer.")
            continue

        # Choose category from given list
        print("\nSelect Category from list:")
        for idx, cat in enumerate(AVAILABLE_CATEGORIES, 1):
            print(f"  {idx}. {cat}")
        
        cat_choice = input(f"Enter choice (1-{len(AVAILABLE_CATEGORIES)}) or type category name: ").strip()
        if not cat_choice:
            break

        if cat_choice.isdigit() and 1 <= int(cat_choice) <= len(AVAILABLE_CATEGORIES):
            category = AVAILABLE_CATEGORIES[int(cat_choice) - 1]
        else:
            category = cat_choice.strip().title()

        add_product(inventory, prod_name, price, qty, category)

    print(f"\n--- Final Inventory for {shopkeeper} ---")
    display_inventory(inventory, shopkeeper)

    # Demonstrate updating stock and price (fulfilling assignment requirement)
    print("\n--- Stock & Price Update Demonstration ---")
    print("Demonstrating update_stock() and update_price() functions:")
    update_stock(inventory, "Milk", 25)
    update_price(inventory, "Apple", 35.0)
    display_inventory(inventory, shopkeeper)

# Run interactive shopkeeper portal
shopkeeper_portal()
```

## Scenario 03: Event Registration and Attendance Management

A university society organizes an event and needs to manage registrations and attendance.

**Your task:**
- Accept participant names and their registration numbers.
- Use a **dictionary** to associate each registration number with a participant's name.
- Maintain a **set** of registered participants to prevent duplicate registrations.
- Store event details, such as event name, date, and venue, in a **tuple**.
- Use a **list** to record participants who attend the event.
- Display registered participants, attendees, absentees, and total attendance percentage.
- Ensure that unregistered participants cannot be marked as present.

```python
# ─────────────────────────────────────────────────────────────────────────────
# Scenario 03 – Event Registration and Attendance Management
# Data structures used:
#   • tuple       → immutable event details (event_name, date, venue)
#   • dictionary  → reg_number → participant_name
#   • set         → registered reg_numbers (fast duplicate check)
#   • list        → reg_numbers of participants who actually attended
# ─────────────────────────────────────────────────────────────────────────────

# TUPLE – event details are fixed/immutable once announced
event_details = ("TechFest 2026", "2026-11-10", "Main Auditorium")
event_name, event_date, event_venue = event_details  # unpack tuple

# Pre-seeded registrations so the event has initial registered participants
registered_participants = {
    "REG001": "Imran",
    "REG002": "Omer",
    "REG003": "Haris",
}

# SET – registered registration numbers (duplicate guard)
registered_reg_numbers = set(registered_participants.keys())

# LIST – registration numbers of participants who attended
attendees_list = []

def register_participant(reg_num, name):
    """Registers a participant. Prevents duplicates using a SET."""
    reg_num = reg_num.strip().upper()
    name    = name.strip()

    if not reg_num or not name:
        print("❌ Registration number and name cannot be empty.")
        return False

    if reg_num in registered_reg_numbers:          # SET lookup – O(1)
        print(f"⚠  '{reg_num}' is already registered to '{registered_participants[reg_num]}' – duplicate rejected.")
        return False

    registered_reg_numbers.add(reg_num)            # add to SET
    registered_participants[reg_num] = name        # add to DICTIONARY
    print(f"✔ Successfully registered: {name} [{reg_num}]")
    return True

def mark_attendance(reg_num):
    """Marks attendance. Unregistered participants cannot be marked present."""
    reg_num = reg_num.strip().upper()

    if reg_num not in registered_reg_numbers:      # SET membership check
        print(f"❌ Error: '{reg_num}' is NOT registered – cannot mark present.")
        return False

    if reg_num in attendees_list:                  # already in LIST
        print(f"⚠  '{reg_num}' ({registered_participants[reg_num]}) is already marked present.")
        return False

    attendees_list.append(reg_num)                 # append to LIST
    print(f"✔ Marked Present: {registered_participants[reg_num]} [{reg_num}]")
    return True

def display_event_report():
    """Generates the full attendance report."""
    total_registered = len(registered_participants)
    total_attended   = len(attendees_list)
    absentees        = [reg for reg in registered_reg_numbers if reg not in attendees_list]

    attendance_pct = (total_attended / total_registered * 100) if total_registered > 0 else 0

    print("\n" + "="*55)
    print(f" Event  : {event_name}")
    print(f" Date   : {event_date}")
    print(f" Venue  : {event_venue}")
    print("="*55)

    print("\n📋 Registered Participants (DICTIONARY):")
    for reg, name in registered_participants.items():
        print(f"   {reg:<8} – {name}")

    print("\n✅ Attendees Present (LIST):")
    if attendees_list:
        for reg in attendees_list:
            print(f"   {reg:<8} – {registered_participants[reg]}")
    else:
        print("   No attendees marked present yet.")

    print("\n❌ Absentees:")
    if absentees:
        for reg in sorted(absentees):
            print(f"   {reg:<8} – {registered_participants[reg]}")
    else:
        print("   None – full attendance!")

    print(f"\n📊 Total Attendance: {total_attended}/{total_registered} ({attendance_pct:.1f}%)")
    print("="*55)

def event_portal():
    """Interactive portal for event registration and attendance."""
    print("="*55)
    print(f" Welcome to {event_name} Management Portal")
    print("="*55)
    print(f"Date: {event_date} | Venue: {event_venue}")

    # --- Phase 1: Interactive Registration ---
    print("\n--- Phase 1: Register New Participants ---")
    print("(Enter participant details. Type 'done' or press Enter/Escape on Reg No to finish)")

    while True:
        reg_num = input("\nEnter Registration Number: ").strip()
        if not reg_num or reg_num.upper() == "DONE":
            break

        name = input(f"Enter Participant Name for '{reg_num}': ").strip()
        if not name:
            break

        register_participant(reg_num, name)

    # --- Phase 2: Interactive Attendance ---
    print("\n--- Phase 2: Mark Attendance ---")
    print("(Enter Reg No to mark present. Type 'done' or press Enter/Escape when finished)")

    while True:
        reg_num = input("\nEnter Reg Number to mark present: ").strip()
        if not reg_num or reg_num.upper() == "DONE":
            break

        mark_attendance(reg_num)

    # --- Phase 3: Final Report ---
    display_event_report()

# Run the interactive event portal
event_portal()
```

## Scenario 04: Online Examination Result Analyzer

A teacher wants to analyze examination results for a class of students across multiple subjects.

**Your task:**
- Accept students' names and their marks in multiple subjects.
- Use a **dictionary** to store each student's subject-wise marks.
- Use a **list** to maintain the class roster.
- Use a **tuple** to store the names of subjects included in the examination.
- Use a **set** to identify the distinct grades achieved by the class, using these grade boundaries: A (80–100), B (70–79), C (60–69), D (50–59), and F (below 50).
- Calculate each student's total marks, average marks, and grade.
- Display the class topper, students who failed in at least one subject, and the class average.
- Validate that marks are between 0 and 100.

```python
# ─────────────────────────────────────────────────────────────────────────────
# Scenario 04 – Online Examination Result Analyzer
# Data structures used:
#   • tuple       → immutable ordered tuple of subject names
#   • list        → class roster (ordered list of student names)
#   • dictionary  → student_name → {subject: marks}
#   • set         → distinct grades achieved across the class
# ─────────────────────────────────────────────────────────────────────────────

# TUPLE – subjects included in the examination (fixed & immutable)
subjects = ("Mathematics", "Physics", "Chemistry", "English", "Computer Science")

# LIST – class roster initially empty, populated entirely by user input
class_roster = []

# DICTIONARY – student marks initially empty, populated entirely by user input
student_marks = {}

def get_grade(average):
    """Return letter grade based on official boundary criteria."""
    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"

def validate_mark(mark):
    """Validate that marks are between 0 and 100."""
    return 0 <= mark <= 100

def accept_students_and_marks():
    """Accept students' names and their marks in multiple subjects directly from the user."""
    print("="*65)
    print(" Online Examination Result Analyzer")
    print(" Examination Subjects (TUPLE):")
    for idx, sub in enumerate(subjects, 1):
        print(f"   {idx}. {sub}")
    print("="*65)
    print("Enter student details below.")
    print("(Type 'done' or press Enter/Escape on Student Name when finished)\n")

    while True:
        student_name = input("Enter Student Name: ").strip()
        if not student_name or student_name.upper() == "DONE":
            break

        clean_name = student_name.title()

        if clean_name not in class_roster:
            class_roster.append(clean_name)     # maintain class roster in a LIST

        marks_dict = {}
        print(f"\nEnter marks for {clean_name} across all {len(subjects)} subjects (0 to 100):")
        escaped = False

        for sub in subjects:                    # iterate through subjects TUPLE
            while True:
                user_val = input(f"  {sub}: ").strip()
                if not user_val or user_val.upper() == "DONE":
                    escaped = True
                    break
                try:
                    mark = float(user_val)
                    if not validate_mark(mark):
                        print("  ❌ Mark must be between 0 and 100. Try again.")
                        continue
                    marks_dict[sub] = mark
                    break
                except ValueError:
                    print("  ❌ Invalid input. Please enter a numerical mark.")

            if escaped:
                break

        if escaped:
            print(f"⚠ Entry cancelled for {clean_name}.")
            if clean_name in class_roster and clean_name not in student_marks:
                class_roster.remove(clean_name)
            break

        # Store in DICTIONARY: student_name -> {subject: marks}
        student_marks[clean_name] = marks_dict
        print(f"✔ Successfully saved marks for {clean_name}!\n")

def analyze_results():
    """Calculate totals, averages, grades, topper, failed students, and class statistics."""
    if not class_roster:
        print("\n⚠ No student records were entered.")
        return

    distinct_grades = set()   # SET – to identify distinct grades achieved
    results = {}
    all_averages = []

    print("\n" + "="*82)
    print(f" {'Student':<12}", end="")
    for sub in subjects:
        print(f" {sub[:6]:>7}", end="")
    print(f" {'Total':>8} {'Average':>9} {'Grade':>6}")
    print("-" * 82)

    for student in class_roster:                  # iterate LIST (class roster)
        marks = student_marks.get(student, {})    # lookup DICTIONARY

        subject_marks_ordered = [marks.get(sub, 0.0) for sub in subjects]
        total   = sum(subject_marks_ordered)
        average = total / len(subjects)
        grade   = get_grade(average)

        distinct_grades.add(grade)                # add to SET
        all_averages.append(average)
        results[student] = {"total": total, "average": average, "grade": grade, "marks": marks}

        print(f" {student:<12}", end="")
        for m in subject_marks_ordered:
            print(f" {m:>7.1f}", end="")
        print(f" {total:>8.1f} {average:>9.2f} {grade:>6}")

    print("=" * 82)

    # 1. Class Topper
    topper = max(results, key=lambda s: results[s]["average"])
    print(f"\n🏆 Class Topper : {topper} (Average: {results[topper]['average']:.2f}, Grade: {results[topper]['grade']})")

    # 2. Students who failed in at least one subject (< 50)
    failed_students = [
        s for s in results
        if any(m < 50 for m in results[s]["marks"].values())
    ]
    print(f"❌ Students who failed in at least one subject: {failed_students if failed_students else 'None'}")

    # 3. Class Average
    class_avg = sum(all_averages) / len(all_averages) if all_averages else 0
    print(f"📊 Overall Class Average : {class_avg:.2f}")

    # 4. Distinct Grades achieved across the class (SET)
    print(f"🎓 Distinct Grades Achieved (SET): {sorted(distinct_grades)}")

# ── Execute ──────────────────────────────────────────────────────────────────
accept_students_and_marks()
analyze_results()
```

## Scenario 05: Personal Expense Tracker and Spending Analysis

A student wants to track daily expenses and identify where their money is being spent.

**Your task:**
- Accept expense descriptions, categories, and amounts over multiple entries.
- Use a **list** to maintain the expense records.
- Use a **tuple** to represent each expense record, containing its description, category, and amount.
- Use a **dictionary** to calculate total expenditure for each category.
- Use a **set** to display all unique expense categories.
- Allow the student to search for expenses by category.
- Display total expenditure, category-wise spending, the highest individual expense, and the category with the highest total spending.
- Reject negative expense amounts and handle cases where no expenses have been recorded.

```python
# ─────────────────────────────────────────────────────────────────────────────
# Scenario 05 – Personal Expense Tracker and Spending Analysis
# Data structures used:
#   • list        → expense_records: maintains ordered log of expense tuples
#   • tuple       → (description, category, amount) immutable record per expense
#   • dictionary  → category → total_spent (expenditure calculated per category)
#   • set         → unique_categories (distinct categories seen)
# ─────────────────────────────────────────────────────────────────────────────

# LIST to maintain expense records (each element is a TUPLE)
expense_records = []

# SET of unique categories
unique_categories = set()

# DICTIONARY: category → total expenditure in PKR
category_totals = {}

def add_expense(description, category, amount):
    """
    Validate and store an expense record.
    Rejects negative expense amounts.
    """
    if amount < 0:
        print(f"❌ Rejected: Negative amount (PKR {amount:,.2f}) for '{description}'. Amounts must be ≥ 0.")
        return False

    clean_desc = description.strip().title()
    clean_cat  = category.strip().title()

    # Store in a TUPLE
    record = (clean_desc, clean_cat, amount)

    # Append to LIST
    expense_records.append(record)

    # Maintain unique categories in a SET
    unique_categories.add(clean_cat)

    # Calculate total expenditure per category in a DICTIONARY
    category_totals[clean_cat] = category_totals.get(clean_cat, 0.0) + amount

    print(f"✔ Recorded: '{clean_desc}' | {clean_cat} | PKR {amount:,.2f}")
    return True

def search_by_category(category_query):
    """Allow the student to search for expenses by category."""
    clean_query = category_query.strip().title()
    matched = [rec for rec in expense_records if rec[1] == clean_query]

    print(f"\n🔍 Search Results for Category: '{clean_query}'")
    if not matched:
        print(f"   No expenses found under '{clean_query}'.")
        return

    print(f"   {'Description':<28} {'Amount (PKR)':>14}")
    print("   " + "-"*44)
    for desc, cat, amt in matched:
        print(f"   {desc:<28} PKR {amt:>10.2f}")
    print("   " + "-"*44)
    print(f"   {'Category Total':<28} PKR {category_totals.get(clean_query, 0.0):>10.2f}")

def display_analysis():
    """Display total expenditure, category-wise spending, highest expense, and top category."""
    # Requirement: Handle cases where no expenses have been recorded
    if not expense_records:
        print("\n" + "="*62)
        print("  PERSONAL EXPENSE TRACKER REPORT")
        print("="*62)
        print("⚠ No expenses have been recorded yet.")
        print("="*62)
        return

    total_expenditure = sum(amt for _, _, amt in expense_records)
    # Highest individual expense from the LIST of TUPLEs
    highest_expense = max(expense_records, key=lambda x: x[2])
    # Category with highest total spending from the DICTIONARY
    highest_spending_category = max(category_totals, key=category_totals.get)

    print("\n" + "="*62)
    print("  PERSONAL EXPENSE TRACKER - SPENDING ANALYSIS")
    print("="*62)
    print(f" {'Description':<26} {'Category':<16} {'Amount (PKR)':>16}")
    print("-" * 62)
    for desc, cat, amt in expense_records:         # iterate LIST
        print(f" {desc:<26} {cat:<16} PKR {amt:>12.2f}")
    print("=" * 62)

    print("\n📊 Category-Wise Spending (DICTIONARY):")
    for cat, total in sorted(category_totals.items(), key=lambda item: -item[1]):
        print(f"   {cat:<18} : PKR {total:>10.2f}")

    print(f"\n🗂  Unique Categories (SET)   : {unique_categories}")
    print(f"💰 Total Expenditure         : PKR {total_expenditure:>10.2f}")
    print(f"📌 Highest Individual Expense: '{highest_expense[0]}' – PKR {highest_expense[2]:,.2f} ({highest_expense[1]})")
    print(f"🏆 Top Spending Category     : {highest_spending_category} (PKR {category_totals[highest_spending_category]:,.2f})")
    print("="*62)

def expense_tracker_portal():
    """Interactive portal accepting expense entries and search queries from the user."""
    print("="*62)
    print(" Student Personal Expense Tracker (PKR)")
    print("="*62)
    print("Enter daily expenses below.")
    print("(Type 'done' or press Enter/Escape on description when finished)\n")

    # Step 1: Accept multiple entries from the user
    while True:
        desc = input("Enter Expense Description: ").strip()
        if not desc or desc.upper() == "DONE":
            break

        cat = input(f"Enter Category for '{desc}': ").strip()
        if not cat or cat.upper() == "DONE":
            break

        while True:
            amt_str = input(f"Enter Amount in PKR for '{desc}': ").strip()
            if not amt_str or amt_str.upper() == "DONE":
                break
            try:
                amt = float(amt_str)
                # Validation: Reject negative expense amounts
                if amt < 0:
                    print("  ❌ Negative amounts not allowed. Must be ≥ 0. Try again.")
                    continue
                add_expense(desc, cat, amt)
                print()
                break
            except ValueError:
                print("  ❌ Invalid amount. Please enter a valid numerical value.")

    # Step 2: Display overall analysis & handle empty case
    display_analysis()

    # Step 3: Interactive Search by category
    if expense_records:
        print("\n--- Search Expenses by Category ---")
        search_cat = input("Enter Category to search (or press Enter/Escape to skip): ").strip()
        if search_cat and search_cat.upper() != "DONE":
            search_by_category(search_cat)

# Run interactive tracker
expense_tracker_portal()
```
